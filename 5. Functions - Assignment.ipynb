```python
def oddEven():
    Input = int(input('Enter an number'))
    if Input % 2 ==0:
        print(Input, 'is even number')
    else:
        print(Input, 'is odd number')
    
    
```

```python
def Marks():
    Total = 0
    for i in range(1,6):
        Mark = int(input(f"Subject {i}: Enter the Mark: "))
        Total = Total + Mark
    return Total
   
```

```python
Totals = Marks()
print('Total mark is : ', Totals)
print('Percentage is : ', float(Totals/5))
```

```python
def Areatriange():
    Height = int (input('Height : '))
    Breadth = int (input('Breadth : '))
    Area = float ((Height * Breadth)/2)
    return Area

def Perimeter():
    Height = int (input('Height 1 : '))
    Height2 = int (input('Height 2 : '))
    Breadth = int (input('Breadth : '))
    Perimeter = Height+Height2+Breadth
    return Perimeter
    
```

```python
print('Area of Triange :', Areatriange())
print('Perimeter of Trinagle :', Perimeter())
```
